#Local Climatological Data (LCD)

```python
import pandas as pd
import glob
import os

from google.colab import drive, userdata
drive.mount('/content/drive')
```

```python
folder_path = '/content/drive/MyDrive/ddds-cohort-21/Projects/Capstone/Data/Local Climatological Data (LCD)'

#Use glob to find all files ending in .csv inside folder
csv_files = glob.glob(os.path.join(folder_path, "*.csv"))

#Read each CSV into a list of DataFrames
df_list = []
for file in csv_files:
    # Read the current CSV file
    df = pd.read_csv(file)
    df_list.append(df)

#Combine all individual DataFrames into one large DataFrame
combined_lcd = pd.concat(df_list, ignore_index=True)
combined_lcd.head()
```

```python
combined_lcd.shape
```

```python
combined_lcd.info()
```

```python
#Convert all mixed-type 'object' columns to strings so Parquet can save them
for col in combined_lcd.select_dtypes(include=['object']).columns:
    combined_lcd[col] = combined_lcd[col].astype(str)
```

#Save as a Parquet

```python
parquet_file = 'combined_lcd.parquet'
parquet_file
```

```python
combined_lcd.to_parquet(parquet_file, index=False)
```
