# Python Exercises - Granger's Elf Analysis

Dr. Granger is interested in studying the relationship between the length of house-elves’ ears and aspects of their DNA. This research is part of a larger project attempting to understand why house-elves possess such powerful magic. She has obtained DNA samples and ear measurements from a small group of house-elves to conduct a preliminary analysis (prior to submitting a grant application to the Ministry of Magic) and she would like you to conduct the analysis for her. She has placed the file on the web for you to download, located ==> [here](https://nyu-cds.github.io/courses/data/houseelf_earlength_dna_data.csv) <==.

You might be able to do this analysis by hand in Excel, but counting all of those bases would be a lot of work, and besides, Dr. Granger seems to always get funded, which means that you’ll be doing this again soon with a much larger dataset. So, you decide to write a script so that it will be easy to do the analysis again.

**BONUS**: Write functions to perform each step

[Source for problem](https://nyu-cds.github.io/courses/exercises/Basic-python-2/)


Loop over rows in `DataFrame`
- Check to see if the ear length is large (>10 cm) or small (<=10 cm)
- Determine the GC-content of the DNA sequence
(i.e., the percentage of bases that are either G or C)
- Store information in **New** `DataFrame`
  - 1st column: ID
  - 2nd column: Ear Size ‘large’ or ‘small’
  - 3rd column: GC Content of DNA sequence
- Print the average GC-content for  large-eared and small-eared elves.
- Exports the table of individual level GC values to a CSV (comma delimited text) file titled grangers_analysis.csv.

```python
import pandas as pd
```

Import into `DataFrame`

```python
url = 'https://nyu-cds.github.io/courses/data/houseelf_earlength_dna_data.csv'
url
```

```python
df = pd.read_csv(url)
df.head(1)
```

# Solution - variant 1

```python
#Create empty lists to hold the data for new DataFrame
id = []
ear_size = []
gc_content = []

#Loop over rows in original DataFrame
for index, row in df.iterrows():

  #1st column: ID
  id.append(row['id'])

  #2nd column: Ear size 'large' or 'small'
  if row['earlength'] > 10:
    ear_size.append('large')
  else:
    ear_size.append('small')

  #3rd column: GC Content of DNA sequence
  dna = row['dnaseq']
  gc = ((dna.count('G') + dna.count('C')) / len(dna)) * 100
  gc_content.append(gc)

#Store in new DataFrame
results_df = pd.DataFrame({
  'ID': id,
  'Ear Size': ear_size,
  'GC Content': gc_content
})

print(results_df)

#Print avg gc-content
averages = results_df.groupby('Ear Size')['GC Content'].mean()

print('Average GC Content')
print('------------------')
for size, avg in averages.items():
  print(f'{size.capitalize()} Ears: {avg:.2f}%')

#Export to new csv
results_df.to_csv('grangers_analysis.csv', index=False)
print("\nData successfully exported to 'grangers_analysis.csv'")
```

# Solution - variant 2

```python
#DEFINE ALL FUNCTIONS

#Check ear length
def ear_size(earlength):
  if earlength > 10:
    return 'large'
  else:
    return 'small'

#Determine GC content of DNA sequence
def gc_content(dnaseq):
  gc_count = dnaseq.count('G') + dnaseq.count('C')
  return(gc_count / len(dnaseq)) * 100

#Loop over DataFrame
def data(df):
  #Store in new DataFrame
  results = pd.DataFrame()

  #Use .apply() to run functions across columns efficiently
  results['ID'] = df['id']                                   #Column 1
  results['Ear Size'] = df['earlength'].apply(ear_size)      #Column 2
  results['GC Content'] = df['dnaseq'].apply(gc_content)     #Column 3

  return results

#Average GC content
def avg_gc(results_df):
  averages = results_df.groupby('Ear Size')['GC Content'].mean()

  print('Average GC Content')
  print('------------------')
  for size, avg in averages.items():
    print(f'{size.capitalize()} Ears: {avg:.2f}%')

#Export results
def export_results(results_df, filename):
  results_df.to_csv(filename, index=False) #Prevents pandas from including row numbers on csv
  print(f"\nResults exported to '{filename}'")


#EXECUTE WORKFLOW

#Process data
results_df = data(df)
print(results_df)
print()

#Print result averages
avg_gc(results_df)

#Export table to csv
export_results(results_df, "grangers_elf_analysis.csv")
```

# Solution - variant 3


```python
def process_elf_data(df, filename='grangers_elf_analysis.csv'):
  '''
  Processes the elf dataset.
  Calculates Ear Size and GC Content.
  Prints averages.
  Exports results to CSV.
  '''

  #Initialize new DataFrame
  results_df = pd.DataFrame()
  results_df['ID'] = df['id']

  #Calculate Ear Size
  results_df['Ear Size'] = df['earlength'].apply(
    lambda x: 'large' if x > 10 else 'small'
  )

  #Calculate GC Content
  results_df['GC Content'] = df['dnaseq'].apply(
    lambda dna: (dna.count('G') + dna.count('C')) / len(dna) * 100
  )

  #Print new DataFrame
  print(results_df)
  print()

  #Calculate and print averages
  averages = results_df.groupby('Ear Size')['GC Content'].mean()
  print('Average GC Content')
  print('------------------')
  for size, avg in averages.items():
    print(f'{size.capitalize()} Ears: {avg:.2f}%')

  #Export to CSV
  results_df.to_csv(filename, index=False)
  print(f"\nResults exported to '{filename}'")

  return results_df

#Execute workflow
final_df = process_elf_data(df)
```
